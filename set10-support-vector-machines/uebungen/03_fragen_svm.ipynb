# Übung 3 – Verständnisfragen zu SVM und SVR

Beantworte die folgenden Fragen in vollständigen Sätzen. Begründe deine Antwort, wenn eine Frage nach einer Wirkung, einer Entscheidung oder einem Vergleich fragt.

## Grundidee, Margin und Support-Vektoren

1. Was versteht man bei einer Support Vector Machine unter einer Hyperebene? Beschreibe, wie eine Hyperebene in einem zwei- und in einem dreidimensionalen Merkmalsraum aussieht.
2. Was bezeichnet der Begriff **Margin** bei einer Support Vector Machine?
3. Warum versucht eine SVM, zwischen mehreren möglichen Entscheidungsgrenzen diejenige mit der größtmöglichen Margin zu finden?
4. Was sind Support-Vektoren, und welchen Einfluss haben sie auf die Lage der Entscheidungsgrenze?
5. Ein Trainingspunkt liegt weit von der Entscheidungsgrenze und der Margin entfernt. Was geschieht normalerweise mit der Entscheidungsgrenze, wenn dieser Punkt geringfügig verschoben wird? Begründe deine Antwort.
6. Worin unterscheiden sich eine Hard-Margin-SVM und eine Soft-Margin-SVM? Für welche Art von Daten ist die Soft-Margin-Variante in der Regel besser geeignet?
7. Welche Bedeutung haben Slack-Variablen bei einer Soft-Margin-SVM?
8. Welche Beobachtungen werden durch den Hinge Loss bestraft, und warum ist diese Verlustfunktion für SVMs geeignet?

## Skalierung und Hyperparameter

9. Warum sollten numerische Merkmale vor dem Training einer SVM normalerweise standardisiert werden? Beschreibe, was ohne Skalierung passieren kann.
10. Wie verändert sich das Verhalten einer SVM bei einem kleinen beziehungsweise einem großen Wert für `C`?
11. Führt ein größerer Wert für `C` automatisch zu einem besseren Modell? Begründe deine Antwort mit Blick auf Underfitting und Overfitting.
12. Welche Bedeutung hat der Hyperparameter `gamma` beim RBF-Kernel? Erkläre die Wirkung eines kleinen und eines großen Wertes anschaulich.
13. Welches Problem kann entstehen, wenn bei einer RBF-SVM sowohl `C` als auch `gamma` sehr groß gewählt werden?
14. Warum beeinflusst der Parameter `gamma` einen linearen Kernel nicht?
15. Was bewirkt die Einstellung `gamma="scale"` in scikit-learn, und warum eignet sie sich häufig als Startwert?
16. Warum sollte der `StandardScaler` innerhalb einer scikit-learn-Pipeline stehen, wenn die Hyperparameter mit Cross-Validation optimiert werden?

## Kernels und Mehrklassenprobleme

17. Was ist die Grundidee des Kernel-Tricks, und welchen Vorteil bietet er bei nichtlinear trennbaren Daten?
18. In welchen Situationen ist ein linearer Kernel ein sinnvoller erster Versuch?
19. Welche zusätzlichen Hyperparameter sind beim polynomialen Kernel wichtig, und was steuert insbesondere `degree`?
20. Vergleiche den linearen, den polynomialen und den RBF-Kernel hinsichtlich Flexibilität, typischer Einsatzgebiete und Interpretierbarkeit.
21. Eine SVM ist zunächst für zwei Klassen definiert. Wie kann `SVC` trotzdem ein Problem mit mehr als zwei Klassen lösen?
22. Worin unterscheiden sich die Mehrklassenstrategien One-vs-Rest und One-vs-One?

## Support Vector Regression

23. Was versteht man bei einer Support Vector Regression unter dem $\varepsilon$-Tunnel?
24. Wie behandelt eine SVR Vorhersagefehler, die vollständig innerhalb des $\varepsilon$-Tunnels liegen?
25. Wie wirkt sich ein größerer Wert für `epsilon` typischerweise auf die Anzahl der Support-Vektoren und auf die Vorhersagegenauigkeit aus?
26. Welche unterschiedlichen Aufgaben erfüllen die Hyperparameter `C` und `epsilon` bei einer SVR?
27. Warum ist Accuracy keine geeignete Metrik, um ein Regressionsmodell zu bewerten?
28. Worin unterscheiden sich MAE und RMSE? Welche der beiden Metriken reagiert stärker auf einzelne große Vorhersagefehler, und warum?

## Modellauswahl und Transfer

29. Warum darf der Testdatensatz nicht für die Auswahl von Kernel und Hyperparametern verwendet werden?
30. Welchen Zweck erfüllt die Cross-Validation innerhalb einer `GridSearchCV`?
31. Warum heißt die RMSE-Metrik in scikit-learn `neg_root_mean_squared_error`, obwohl ein kleiner RMSE eigentlich besser ist?
32. Ein Modell erreicht einen F1-Score von 1,00 auf den Trainingsdaten, 0,72 in der Cross-Validation und 0,69 auf den Testdaten. Wie würdest du dieses Ergebnis diagnostizieren, und welche Gegenmaßnahmen wären denkbar?
33. Eine RBF-SVM sagt für fast alle Beobachtungen dieselbe Klasse voraus. Welche Einstellungen von `C` oder `gamma` könnten dieses starke Underfitting verursacht haben?
34. Nach einer Vergrößerung von `epsilon` verwendet eine SVR deutlich weniger Support-Vektoren, erreicht aber einen höheren Test-RMSE. Wie lässt sich dieses Ergebnis erklären?
35. Nenne jeweils zwei Vorteile und zwei Nachteile von Support Vector Machines.
36. Warum können Kernel-SVMs bei sehr großen Datensätzen hinsichtlich Laufzeit und Speicherbedarf problematisch werden?
37. Du möchtest Textdokumente mit 50.000 dünn besetzten Merkmalen klassifizieren. Welchen Kernel würdest du zuerst ausprobieren, und warum?
38. Ein zweidimensionaler Datensatz besitzt ungefähr eine kreisförmige Klassengrenze. Welcher Kernel wäre ein plausibler Ausgangspunkt, und warum?
39. In einer Anwendung ist es besonders wichtig, möglichst viele Beobachtungen einer seltenen positiven Klasse zu erkennen. Welche Bewertungsmetrik und welche Einstellung von `SVC` solltest du besonders beachten?
40. Beschreibe einen sauberen ML-Arbeitsablauf für eine SVM: Beginne bei den Rohdaten und ende bei der einmaligen finalen Bewertung auf dem Testdatensatz.
