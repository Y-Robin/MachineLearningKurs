## 1. Variablen und grundlegende Datentypen

```python
kurs = "Machine Learning"   # str
anzahl_teilnehmende = 18       # int
lernrate = 0.01                # float
ist_aktiv = True               # bool

print(kurs, anzahl_teilnehmende, lernrate, ist_aktiv)
print(type(lernrate))
```

Python ist dynamisch typisiert: Der Typ gehört zum Wert und muss nicht vorab deklariert werden. Sinnvolle Variablennamen sind wichtiger als kurze Namen. Kommentare erklären das **Warum**, nicht jede offensichtliche Zeile.

## 2. Listen, Tupel, Dictionaries und Sets

```python
messwerte = [12.1, 11.8, 12.5, 13.0]      # geordnet, veränderbar
form = (28, 28)                            # geordnet, unveränderbar
person = {"name": "Ada", "alter": 36}  # Schlüssel -> Wert
klassen = {"Katze", "Hund", "Katze"}    # eindeutige Werte

messwerte.append(13.2)
print(messwerte[0], messwerte[-1])
print(messwerte[1:4])
print(person["name"])
print(klassen)
```

## 3. Bedingungen und Schleifen

```python
temperatur = 21.5

if temperatur >= 25:
    bewertung = "warm"
elif temperatur >= 15:
    bewertung = "mild"
else:
    bewertung = "kühl"

print(bewertung)

for index, wert in enumerate(messwerte):
    print(f"Messung {index + 1}: {wert:.1f}")
```

## 4. Funktionen und List Comprehensions

```python
def celsius_zu_fahrenheit(celsius: float) -> float:
    """Wandelt eine Temperatur von Celsius in Fahrenheit um."""
    return celsius * 9 / 5 + 32

fahrenheit = [celsius_zu_fahrenheit(wert) for wert in messwerte]
print(fahrenheit)
```

Type Hints wie `float` dokumentieren erwartete Typen, werden von Python aber nicht automatisch erzwungen. Funktionen sollten möglichst eine klar abgegrenzte Aufgabe erfüllen.

## 5. NumPy: Arrays statt Python-Listen

NumPy-Arrays enthalten normalerweise Werte eines gemeinsamen Datentyps. Sie erlauben schnelle, vektorisierte Berechnungen über ganze Datenblöcke und bilden die Grundlage vieler ML-Bibliotheken.

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5], dtype=float)
matrix = np.array([[1, 2, 3], [4, 5, 6]])

print("x:     ", x)
print("Form:  ", matrix.shape)
print("Dim.:  ", matrix.ndim)
print("Typ:   ", matrix.dtype)
```

### Vektorisierung: Rechnen ohne explizite Schleife

```python
print("x * 10:  ", x * 10)
print("x²:      ", x**2)
print("Mittel:  ", x.mean())
print("Std.abw.:", x.std())
print("Summe:   ", x.sum())
```

### Indexierung, Slicing und boolesche Masken

```python
daten = np.array([10, 15, 8, 22, 17, 5])

print(daten[1:4])
maske = daten >= 15
print("Maske: ", maske)
print("Gefiltert:", daten[maske])
```

### Achsen, Reshaping und Broadcasting

```python
a = np.arange(1, 7).reshape(2, 3)
spalten_offset = np.array([10, 20, 30])

print("Matrix:\n", a)
print("Spaltensummen:", a.sum(axis=0))
print("Zeilensummen:  ", a.sum(axis=1))
print("Broadcasting:\n", a + spalten_offset)
```

Bei einer Matrix ist `axis=0` die Aggregation **über die Zeilen hinweg** (ein Ergebnis je Spalte), `axis=1` über die Spalten hinweg (ein Ergebnis je Zeile). Broadcasting erweitert kompatible Formen virtuell, ohne die Daten manuell zu kopieren.

### Zufallszahlen reproduzierbar erzeugen

```python
rng = np.random.default_rng(seed=42)
stichprobe = rng.normal(loc=100, scale=15, size=8)
print(np.round(stichprobe, 1))
```

Ein fester Seed macht ein Zufallsexperiment reproduzierbar. Das ist für Unterricht, Tests und Modellvergleiche wichtig.

## Übung

Erzeuge ein NumPy-Array mit den Prüfungswerten `[55, 72, 88, 49, 91, 67]`. Berechne Mittelwert, Minimum und Maximum. Filtere anschließend alle bestandenen Werte (mindestens 60 Punkte).

```python
# Musterlösung
punkte = np.array([55, 72, 88, 49, 91, 67])
print("Mittelwert:", punkte.mean())
print("Minimum:   ", punkte.min())
print("Maximum:   ", punkte.max())
print("Bestanden: ", punkte[punkte >= 60])
```

## Merksätze

- Listen sind flexibel; NumPy-Arrays sind für gleichartige numerische Daten optimiert.
- `shape` beschreibt die Form eines Arrays.
- Vektorisierung ersetzt viele Schleifen durch Operationen auf ganzen Arrays.
- Boolesche Masken filtern Daten.
- Ein fester Seed macht Zufallszahlen reproduzierbar.
