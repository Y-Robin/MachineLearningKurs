## 2. Was sind Bibliotheken?
Python bringt viele Grundfunktionen mit. Bibliotheken liefern zusätzliche Werkzeuge, damit wir nicht alles selbst programmieren müssen.

Beispiele für diesen Kurs:

| Bibliothek | Aufgabe |
|---|---|
| NumPy | Rechnen mit vielen Zahlen |
| pandas | Arbeiten mit Tabellen |
| Matplotlib | Diagramme zeichnen |
| scikit-learn | Machine-Learning-Modelle verwenden |

Eine Bibliothek wird einmal installiert und anschließend im Code importiert.

## 3. Warum eine virtuelle Umgebung?
Eine virtuelle Umgebung ist ein eigener Bereich für die Bibliotheken dieses Kurses. Dadurch geraten Kurs-Pakete nicht mit Paketen anderer Python-Projekte durcheinander.

Kurz gesagt: **Ein Projekt bekommt seine eigene Paketbox.**

## 4. Einmalige Einrichtung
Öffne ein Terminal im Kursordner und führe die Befehle passend zu deinem Betriebssystem aus.

### Windows PowerShell

    py -m venv .venv
    .venv\Scripts\Activate.ps1
    python -m pip install -r requirements.txt
    python -m jupyter lab

### macOS/Linux

    python3 -m venv .venv
    source .venv/bin/activate
    python -m pip install -r requirements.txt
    python -m jupyter lab

Beim nächsten Mal reichen Aktivieren und Starten von JupyterLab.

## 5. Eine Bibliothek importieren
Mit import machen wir die Werkzeuge einer installierten Bibliothek im Notebook verfügbar. Häufig nutzt man kurze, allgemein bekannte Abkürzungen.

```python
import numpy as np

zahlen = np.array([2, 4, 6, 8])
print("Zahlen:", zahlen)
print("Mittelwert:", zahlen.mean())
```

NumPy wurde als **np** abgekürzt. Über den Punkt wählen wir ein Werkzeug der Bibliothek aus, zum Beispiel np.array.

## 6. Ein Werkzeug aus Python selbst
Einige Werkzeuge sind bereits bei Python dabei und müssen nicht extra installiert werden.

```python
import math

print("Quadratwurzel aus 81:", math.sqrt(81))
print("Pi gerundet:", round(math.pi, 2))
```
