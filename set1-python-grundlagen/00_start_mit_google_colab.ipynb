# 00 – Den Kurs mit Google Colab starten

## 1. Notebook öffnen

Öffne ein Notebook auf GitHub und klicke oben auf **Open in Colab**. Verbinde dich mit einer Laufzeit und führe zuerst die Setup-Zelle aus. Mit **Shift + Enter** arbeitest du anschließend von oben nach unten.

Der Button öffnet das Notebook. Die Setup-Zelle lädt zusätzlich das gesamte Repository nach `/content/MachineLearningKurs`, installiert die Kursbibliotheken und setzt den Arbeitsordner passend zum Notebook. So sind auch Daten und Python-Hilfsdateien verfügbar.

## 2. Einrichtung prüfen

Die Setup-Zelle steht direkt oben in jedem Notebook, auch in den Übungen. Jedes geöffnete Notebook benötigt sein eigenes Setup. Falls Colab nach der Installation einen Neustart der Sitzung verlangt, starte sie neu und führe die Setup-Zelle erneut aus.

In lokalem Jupyter überspringt die Zelle das Colab-Setup. Die virtuelle Umgebung und lokale Installation aus Notebook 01 brauchst du in Colab nicht anzulegen.
