# Übung 3 – Verständnisfragen zu neuronalen Netzen

Beantworte die Fragen in vollständigen Sätzen und begründe Entscheidungen oder Vergleiche.

1. Welche Berechnung führt ein einzelnes künstliches Neuron aus?

## Aktivierungen und Loss
