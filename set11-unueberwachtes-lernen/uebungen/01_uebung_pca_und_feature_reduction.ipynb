```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.feature_selection import VarianceThreshold
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_validate, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)

n = 750
latent = rng.normal(size=(n, 4))
X = pd.DataFrame({
    "m01": 10 + 4*latent[:, 0] + rng.normal(0, .3, n),
    "m02": 100 + 38*latent[:, 0] + rng.normal(0, 3, n),
    "m03": -2*latent[:, 0] + 2*latent[:, 1] + rng.normal(0, .4, n),
    "m04": 5*latent[:, 1] + rng.normal(0, .4, n),
    "m05": 4.8*latent[:, 1] + rng.normal(0, .5, n),
    "m06": 3*latent[:, 2] + rng.normal(0, .3, n),
    "m07": -2*latent[:, 2] + 2*latent[:, 3] + rng.normal(0, .4, n),
    "m08": 4*latent[:, 3] + rng.normal(0, .4, n),
    "m09": rng.normal(size=n),
    "m10": rng.normal(size=n),
    "m11": rng.normal(size=n),
    "m12": 1 + rng.normal(0, .001, n),
})
y_demo = (
    latent[:, 0] - .8*latent[:, 1] + .7*latent[:, 2]
    + rng.normal(0, .8, n) > 0
).astype(int)
X_train, X_test, y_train, y_test = train_test_split(
    X, y_demo, test_size=.25, stratify=y_demo, random_state=RANDOM_STATE
)
print("Daten bereit:", X_train.shape, X_test.shape)
```
