# Set 02 – Fallstudien: Welche ML-Methode passt?
Eine praktische Übung ohne Programmierung. Arbeitet in Gruppen von 2–3 Personen etwa 45 Minuten und stellt anschließend einen Fall in fünf Minuten vor.

## Analysevorlage
Beantwortet pro Fall:

1. Was soll das System ausgeben?
2. Welche Eingaben gibt es?
3. Gibt es bekannte Labels?
4. Welche Lernart passt?
5. Ist es Klassifikation, Regression, Clustering oder Anomalieerkennung?
6. Wie wird die Qualität bewertet?
7. Welcher Fehler ist besonders kritisch?
8. Welche Frage muss vor Projektbeginn geklärt werden?

## Fall 1 – Maschinenwartung
Vibration, Temperatur und Stromaufnahme werden erfasst. Es gibt sehr viele Aufzeichnungen aus dem Normalbetrieb, aber nur drei dokumentierte Defekte. Die Instandhaltung möchte früh gewarnt werden.

**Analyse:** Ausgabe / Eingaben / Labels / Lernart und Methode / Bewertung / kritischer Fehler / offene Fragen

## Fall 2 – Hauspreise
Vorhanden sind Wohnfläche, Lage, Baujahr, Zimmeranzahl und historische Verkaufspreise. Für neue Häuser soll ein Preis geschätzt werden.

**Analyse:** Ausgabe / Eingaben / Labels / Lernart und Methode / Bewertung / kritischer Fehler / offene Fragen

## Fall 3 – Kundengruppen
Ein Shop kennt Käufe, Klicks, Umsatz und Besuchshäufigkeit. Vorgegebene Kundentypen gibt es nicht. Ähnliche Kunden sollen gruppiert werden.

**Analyse:** Ausgabe / Eingaben / Labels / Lernart und Methode / Bewertung / kritischer Fehler / offene Fragen

## Fall 4 – E-Mail-Filter
Viele ältere E-Mails sind als Spam oder Nicht-Spam markiert. Neue Nachrichten sollen anhand von Wörtern, Absender, Links und Betreff eingeordnet werden.

**Analyse:** Ausgabe / Eingaben / Labels / Lernart und Methode / Bewertung / kritischer Fehler / offene Fragen

## Fall 5 – Snake
Ein Agent kennt Schlange, Futter und Wände und kann eine Bewegungsrichtung wählen. Er soll lange überleben und viel Futter sammeln.

**Analyse:** Zustand / Aktionen / Belohnung und Strafe / Lernart / Bewertung / offenes Problem

## Fall 6 – Diagnoseunterstützung
Bilder, Laborwerte oder Biosignale sollen auf auffällige Befunde hinweisen. Historische Fälle mit Befund sind vorhanden; Fachpersonal entscheidet endgültig.

**Analyse:** Ausgabe / Eingaben / Labels / Lernart und Methode / Bewertung / kritischster Fehler / Schutzmaßnahmen

## Zusatz – Schwellwert
Das Wartungssystem meldet ab einem Anomaliescore-Schwellwert Alarm. Was passiert wahrscheinlich bei 20 beziehungsweise 95 auf einer Skala von 0 bis 100? Welche Informationen benötigt ihr für eine sinnvolle Wahl und wer sollte beteiligt sein?

## Präsentation
Begründet für einen Fall Methodenwahl, Daten, Bewertung und gefährlichsten Fehler. Nennt außerdem eine offene Frage, die vor einer Umsetzung beantwortet werden muss.
