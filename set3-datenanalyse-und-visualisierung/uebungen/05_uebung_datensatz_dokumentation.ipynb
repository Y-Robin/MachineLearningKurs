3. Falls beim Entpacken ein zusätzlicher Unterordner entsteht, verschiebe die beiden CSVs in den angegebenen Zielordner. Weitere Begleitdateien dürfen dort ebenfalls liegen.
4. Halte die Kaggle-Data-Card und die verlinkte [UCI-Originaldokumentation: Student Performance](https://archive.ics.uci.edu/dataset/320/student+performance) geöffnet. Notiere URL und Abrufdatum. Für die Übung ist keine separate `data_description.txt` nötig.

Die folgende Dateiprüfung prüft nur Ablage und Lesbarkeit. Es gibt keinen automatischen Downloader und keine API-Einrichtung.

````python
from pathlib import Path
import pandas as pd

wurzel = next(p for p in [Path.cwd(), *Path.cwd().parents]
              if (p / 'set3-datenanalyse-und-visualisierung').is_dir())
ordner = wurzel / 'set3-datenanalyse-und-visualisierung/daten/kaggle/student-performance'
alles_ok = True
for name in ['student-mat.csv', 'student-por.csv']:
    try:
        # Unterschiedliche Exportfassungen können Komma oder Semikolon verwenden.
        probe = pd.read_csv(ordner / name, sep=None, engine='python', nrows=3)
        erwartet = {'school', 'age', 'studytime', 'Dalc', 'Walc', 'G1', 'G2', 'G3'}
        if probe.empty or not erwartet.issubset(probe.columns):
            raise ValueError('Leere oder falsche CSV; Dateiname und Spalten prüfen.')
        print('OK:', name, 'ist vorhanden und lesbar.')
    except (OSError, ValueError, UnicodeError, pd.errors.ParserError) as exc:
        alles_ok = False
        print('PRÜFEN:', name, '-', exc)
print('Ablageordner:', ordner)
if alles_ok:
    print('Dateien erfolgreich geprüft. Jetzt die Dokumentation lesen.')
else:
    print('ZIP entpacken, beide CSVs direkt hier ablegen und die Zelle erneut ausführen.')
````
